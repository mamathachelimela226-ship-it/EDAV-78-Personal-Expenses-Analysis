```python
import pandas as pd
from IPython.display import display

path = "/content/drive/MyDrive/PhonePe_Statement_Oct2025_Oct2026.csv"
df_raw = pd.read_csv(path, header=None, dtype=str)
header_row = 0
for i in range(len(df_raw)):
  if "Transaction Details" in str(df_raw.iloc[i].values):
    header_row = i
    break

df = pd.read_csv(path, skiprows=header_row, dtype=str)

df = df.loc[:, ~df.columns.str.contains('^Unnamed', na=False)]

df = df.dropna(how='all')
df = df[~df.astype(str).apply(lambda x: x.str.contains('Disclaimer', na=False)).any(axis=1)]

display(df)

print(f"Rows: {len(df)} | Columns: {list(df.columns)}")
```

```python
import pandas as pd

df = df[df['Transaction Details'].notna()]
df = df[~df['Transaction Details'].astype(str).str.contains('Disclaimer', case=False, na=False)]

df['Date'] = pd.to_datetime(df['Date'], dayfirst=True, errors='coerce')

df['Amount'] = df['Amount'].astype(str).str.replace('₹','').str.replace(',','').str.replace('Rs.','').str.replace('Rs','').str.strip()
df['Amount'] = pd.to_numeric(df['Amount'], errors='coerce')

df_clean = df.dropna(subset=['Amount']).copy()

print(df_clean.dtypes)
print(df_clean['Amount'].describe())

from IPython.display import display
display(df_clean.head(10))

print(f"Rs {df_clean['Amount'].sum():,.2f}")
```

```python
print("=== 1. DATA CLEANING ===")
print("Nulls:\n", df.isnull().sum())
print("Duplicates Before:", df.duplicated().sum())

# Remove duplicates and disclaimer rows
df = df.drop_duplicates()
df = df[df['Transaction Details'].notna()]
df = df[~df['Transaction Details'].astype(str).str.contains('Disclaimer', case=False, na=False)]

# Convert Date and Amount
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
df['Amount'] = df['Amount'].astype(str).str.replace(',','').str.strip()
df['Amount'] = pd.to_numeric(df['Amount'], errors='coerce')
df = df.dropna(subset=['Amount', 'Date'])

# Clean Type and derive Month
df['Transaction Type'] = df['Transaction Type'].astype(str).str.strip().str.upper()
df['Month'] = df['Date'].dt.month_name()
df['Month_Num'] = df['Date'].dt.month
df['Weekday'] = df['Date'].dt.day_name()

print("Duplicates After:", df.duplicated().sum())
print("Rows After Cleaning:", len(df))
```

```python
print("=== 2. DATA VALIDATION ===")
print("Min:", df['Amount'].min(), "Max:", df['Amount'].max())
print("Invalid <=0:", len(df[df['Amount'] <= 0]))

df = df[df['Amount'] > 0]

# Split for analysis
debit_df = df[df['Transaction Type'] == 'DEBIT'].copy()
credit_df = df[df['Transaction Type'] == 'CREDIT'].copy()

print("Final Rows:", len(df))
print(f"DEBIT: {len(debit_df)} rows, CREDIT: {len(credit_df)} rows")
```

```python
print("=== 3. OUTLIER DETECTION (IQR) ===")

Q1 = df['Amount'].quantile(0.25)
Q3 = df['Amount'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR

outliers = df[(df['Amount'] < lower) | (df['Amount'] > upper)]

print(f"Q1: {Q1}, Q3: {Q3}, IQR: {IQR}")
print(f"Bounds: {lower:.2f} to {upper:.2f}")
print(f"Outliers: {len(outliers)} rows")
print(outliers[['Date','Transaction Type','Amount']].head())
print("Note: Outliers kept - PG, Fees valid")
```

```python
print("=== 4. STATISTICAL SUMMARY ===")
print(df['Amount'].describe())
print("\nType Wise Total:\n", df.groupby('Transaction Type')['Amount'].sum())
print(f"\nTotal DEBIT: Rs {debit_df['Amount'].sum():,.2f}")
print(f"Total CREDIT: Rs {credit_df['Amount'].sum():,.2f}")
print("\nMonth Wise DEBIT:\n", debit_df.groupby('Month')['Amount'].sum().sort_values(ascending=False))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
debit_temp = debit_df.copy()
debit_temp['Date'] = pd.to_datetime(debit_temp['Date'])
debit_temp = debit_temp.sort_values('Date')
debit_temp['YearMonth_Num'] = debit_temp['Date'].dt.strftime('%Y-%m')
debit_temp['Month_Name'] = debit_temp['Date'].dt.strftime('%b %Y')  # Oct 2025, Nov 2025...
monthly = debit_temp.groupby(['YearMonth_Num', 'Month_Name'])['Amount'].sum().reset_index()
monthly = monthly.sort_values('YearMonth_Num')

# GRAPH 1: Monthly Trend
plt.figure(figsize=(12,4))
plt.plot(monthly['Month_Name'], monthly['Amount'], marker='o', color='blue', linewidth=2)
plt.title('Monthly Expense Trend - Roll78')
plt.xlabel('Month Name')
plt.ylabel('DEBIT Rs')
plt.xticks(rotation=45)
plt.grid(True)
plt.savefig('1_trend_monthname.png', dpi=300, bbox_inches='tight')
plt.show()

# GRAPH 2: Bar
plt.figure(figsize=(12,4))
plt.bar(monthly['Month_Name'], monthly['Amount'], color='skyblue')
plt.title('Month Wise Expense - Month Names Order - Roll78')
plt.xlabel('Month Name')
plt.ylabel('Amount Rs')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('2_bar_monthname.png', dpi=300, bbox_inches='tight')
plt.show()

# GRAPH 3: Pie
plt.figure(figsize=(6,6))
s = df.groupby('Transaction Type')['Amount'].sum()
plt.pie(s.values, labels=s.index, autopct='%1.1f%%', colors=['#ff9999','#66b3ff'])
plt.title('DEBIT vs CREDIT - Roll78')
plt.savefig('3_pie.png', dpi=300, bbox_inches='tight')
plt.show()

# GRAPH 4: Hist
plt.figure(figsize=(8,4))
sns.histplot(debit_temp['Amount'], bins=30, color='green')
plt.title('Amount Distribution - Roll78')
plt.savefig('4_hist.png', dpi=300, bbox_inches='tight')
plt.show()

# GRAPH 5: Boxplot
plt.figure(figsize=(5,4))
sns.boxplot(y=debit_temp['Amount'], color='orange')
plt.title('Boxplot - Roll78')
plt.savefig('5_box.png', dpi=300, bbox_inches='tight')
plt.show()

# GRAPH 6: Weekday
plt.figure(figsize=(8,4))
order = ['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday']
sns.countplot(data=debit_temp, x='Weekday', order=order)
plt.title('Weekday Transactions - Roll78')
plt.savefig('6_weekday.png', dpi=300, bbox_inches='tight')
plt.show()

print("Done - Month Names Oct 2025, Nov 2025... proper order lo vachayi!")
```

```python
# After graphs - Print Final Analysis for Report

print("=== FINAL ANALYSIS - Roll78 ===\n")

# Total
print(f"Total DEBIT: Rs {debit_temp['Amount'].sum():,.2f}")
print(f"Total CREDIT: Rs {df[df['Transaction Type']=='CREDIT']['Amount'].sum():,.2f}")
print(f"Total Transactions: {len(df)}")

# Month wise with names
print("\nMonth Wise Expense (Month Name Order):")
print(monthly[['Month_Name','Amount']].to_string(index=False))

# Highest & Lowest
print(f"\nHighest Expense Month: {monthly.loc[monthly['Amount'].idxmax(), 'Month_Name']} - Rs {monthly['Amount'].max():,.2f}")
print(f"Lowest Expense Month: {monthly.loc[monthly['Amount'].idxmin(), 'Month_Name']} - Rs {monthly['Amount'].min():,.2f}")

# Weekday
print("\nWeekday Wise Count:")
print(debit_temp['Weekday'].value_counts())

# Outlier
Q1 = df['Amount'].quantile(0.25)
Q3 = df['Amount'].quantile(0.75)
IQR = Q3-Q1
print(f"\nOutlier Bounds: {Q1-1.5*IQR:.2f} to {Q3+1.5*IQR:.2f}")

# Save monthly summary for report
monthly.to_csv('MonthWise_Summary_Roll78.csv', index=False)
print("\nSaved: MonthWise_Summary_Roll78.csv")
```
